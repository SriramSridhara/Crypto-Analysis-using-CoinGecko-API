```python
def update_excel():
    payload = {'vs_currency':'usd', 'order':'market_cap_desc', 'per_page':'50', 'page':'1', 'sparkline':'false', 'price_change_percentage':'24h'}
    base_url = 'https://api.coingecko.com/api/v3/coins/markets'
        
    headers = {
        "accept": "application/json",
        "x-cg-demo-api-key": "CG-KL39BmUZUVjjjNLPvBgHR4DH"
    }
    
    response = requests.get(base_url, params=payload, headers=headers)
    if response.status_code == 200:
        data = response.json()
        
        df = pd.DataFrame([{
            'Name': coin['name'],
            'Symbol': coin['symbol'].upper(),
            'Current Price (USD)': coin['current_price'],
            'Market Cap (USD)': coin['market_cap'],
            '24h Trading Volume (USD)': coin['total_volume'],
            '24h Price Change (%)': float(coin['price_change_percentage_24h'])
        } for coin in data])

        # Open Excel and Update the Sheet
        excel = win32.Dispatch('Excel.Application')
        excel.Visible = True
        wb = excel.Workbooks.Open(r'C:\Users\HP\OneDrive\Desktop\Ram\Projects\CoinGecko API Project.xlsx')
        ws = wb.Sheets("Sheet1")

        ws.Range("A1").Value = "Name"
        ws.Range("B1").Value = "Symbol"
        ws.Range("C1").Value = "Current Price (USD)"
        ws.Range("D1").Value = "Market Cap (USD)"
        ws.Range("E1").Value = "24h Trading Volume (USD)"
        ws.Range("F1").Value = "24h Price Change (%)"

        #variables used for basic data analytics
        top_five_crypto = list()
        price = float(0)
        highest_percent_change = float(-100)
        highest_change_coin = "_"
        lowest_percent_change = float(100)
        lowest_change_coin = "_"
        
        for i, row in df.iterrows():
            #basic data analytics
            price  += row['Current Price (USD)']
            if abs(row['24h Price Change (%)']) > highest_percent_change:
                highest_percent_change = abs(row['24h Price Change (%)'])
                highest_change_coin = row['Name']

            if abs(row['24h Price Change (%)']) < lowest_percent_change:
                lowest_percent_change = abs(row['24h Price Change (%)'])
                lowest_change_coin = row['Name']

            if i<5:
                top_five_crypto.append(row['Name'])
            
            ws.Cells(i+2, 1).Value = row['Name']
            ws.Cells(i+2, 2).Value = row['Symbol']
            ws.Cells(i+2, 3).Value = row['Current Price (USD)']
            ws.Cells(i+2, 4).Value = row['Market Cap (USD)']
            ws.Cells(i+2, 5).Value = row['24h Trading Volume (USD)']
            ws.Cells(i+2, 6).Value = row['24h Price Change (%)']

        wb.Save()        
        print("Excel updated successfully.")

        #print(df.head())
        print(f"The top 5 Cryptocurrencies by Market Cap are : {top_five_crypto}")
        print(f"Average Price of the Top 50 Crypto Currencies is : {price/50}")
        print(f"Highest % change in last 24 hours was observed in : {highest_change_coin} by {highest_percent_change} percent")
        print(f"Lowest % change in last 24 hours was observed in : {lowest_change_coin} by {lowest_percent_change} percent")
    else:
        print("Failed to fetch data:", response.status_code)
```

```python
if __name__ == "__main__":
    while True:
        update_excel()   
        time.sleep(300)  # Wait for 5 min

```
